## Подготовка

```python
from pathlib import Path
import sys
import pandas as pd

print("Python:", sys.executable)
print("pandas:", pd.__version__)
print("Рабочая папка:", Path.cwd())
DATA = Path("../data")
assert DATA.is_dir(), "Рядом с папкой ноутбука должна находиться data/"
```

## Основной набор decisions.csv

```python
home_decisions = pd.read_csv(
    DATA / "decisions.csv", dtype={"lei": "string", "county_code": "string"}
)
home_outcomes = home_decisions["action_taken"]
home_n = len(home_decisions)
assert home_n > 0
assert home_outcomes.isna().sum() == 0
assert home_outcomes.isin([1, 2, 3]).sum() == home_n
print(home_decisions.shape)
home_outcomes.value_counts(dropna=False)
```

## H1. Доля отказов

```python
home_denied = home_outcomes.eq(3).sum()
home_denial_rate = home_denied / home_n
print(f"Отказы {home_denied}/{home_n}: {100 * home_denial_rate:.2f}%")
home_approved = home_outcomes.isin([1, 2]).sum()
assert home_denied + home_approved == home_n
assert 0 <= home_denied <= home_n
assert 0 <= home_denial_rate <= 1
```

## H2. Одобрение без принятия предложения

```python
home_approved_not_accepted = home_outcomes.eq(2).sum()
home_approved_not_accepted_rate = home_approved_not_accepted / home_n
home_approval_rate = home_outcomes.isin([1, 2]).sum() / home_n
home_origination_rate = home_outcomes.eq(1).sum() / home_n
print(f"Одобрено, но не принято {home_approved_not_accepted}/{home_n}: {100 * home_approved_not_accepted_rate:.2f}%")
print(f"Разность одобрения и выдачи: {100 * (home_approval_rate - home_origination_rate):.2f} п.п.")
home_issued = home_outcomes.eq(1).sum()
assert home_approved - home_issued == home_approved_not_accepted
```

## H3. Инструменты

Series - одномерный индексированный массив элементов.
DataFrame - совокупность объектов Series с общим индексом.
nrows = 3 - ограничивает чтение и загружает только первые три строки непосредственно из файла.
head(5) - отображает первые 5 строк из уже загруженной в оперативную память таблицы.
Метод и атрибут: Метод - home_outcomes.eq(3).sum(), Атрибут - home_decisions.shape. Метод производит вычисление/действие, а атрибут возвращает сохраненное свойство структуры.
Ссылка на документацию и детали: https://pandas.pydata.org/docs/reference/api/pandas.Series.isin.html - функция isin принимает перечень значений и выдает логическую Series; применяется для фильтрации исходов 1 и 2 одним выражением.
